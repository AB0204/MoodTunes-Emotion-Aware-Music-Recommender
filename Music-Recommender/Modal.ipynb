# 🔬 MoodTunes: Recommendation Architecture & Model Evaluation

This notebook documents the deep learning architecture behind **MoodTunes**.

## 1. System Overview
MoodTunes utilizes a two-stage pipeline for emotion-based music recommendation:
1. **Computer Vision Stage:** Real-time facial emotion classification using a pre-trained Keras CNN model.
2. **Music Recommendation Stage:** Direct mapping of classified emotions to Spotify audio features (Valence, Energy, Tempo) via the Spotipy API client.

### Why direct mapping over GCN?
While Multi-modal Knowledge Graph Convolutional Networks (MKGCN) were researched for graph-based item embedding, they present significant cold-start constraints, high latency, and heavy computational requirements. For real-time PyQt5 desktop applications, direct mapping to Spotify's dynamic recommendation engine provides a more responsive, reliable, and production-grade solution.
